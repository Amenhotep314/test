# Units

Units will be represented in 3D space, with each direction indicating the exponent associated with that basis vector. The basis vectors are length, mass, and time.

```python
import numpy as np
import plotly.express as px
import matplotlib.pyplot as plt
```

```python
basis = ["length", "mass", "time", "ampere", "kelvin"]
units = {
    # Using only m-kg-s
    "length": (1, 0, 0, 0, 0),
    "mass": (0, 1, 0, 0, 0),
    "time": (0, 0, 1, 0, 0),

    "frequency": (0, 0, -1, 0, 0),
    "angular acceleration": (0, 0, -2, 0, 0),
    "angular momentum": (2, 1, -1, 0, 0),
    "moment of inertia": (2, 1, 0, 0, 0),
    "torque"

    "area": (2, 0, 0, 0, 0),
    "volume": (3, 0, 0, 0, 0),
    "hypervolume": (4, 0, 0, 0, 0),

    "linear mass density": (-1, 1, 0, 0, 0),
    "surface mass density": (-2, 1, 0, 0, 0),
    "mass density": (-3, 1, 0, 0, 0),

    "velocity": (1, 0, -1, 0, 0),
    "acceleration": (1, 0, -2, 0, 0),
    "jerk": (1, 0, -3, 0, 0),

    "momentum": (1, 1, -1, 0, 0),
    "force": (1, 1, -2, 0, 0),
    "energy": (2, 1, -2, 0, 0),
    "power": (2, 1, -3, 0, 0),

    "pressure": (-1, 1, -2, 0, 0),
    "viscosity": (-1, 1, -1, 0, 0),
}
```

```python
# Define the points and their labels
labels = units.keys()
points = [point[:3] for point in units.values()]

# Convert the points into a format suitable for plotting
x, y, z = zip(*points)

# Create the 3D scatter plot
fig = px.scatter_3d(x=x, y=y, z=z, text=labels, title="3D Scatter Plot of Units")
fig.update_traces(marker=dict(size=8))
```
