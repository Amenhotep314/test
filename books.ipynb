```python
plt.hist(years, bins=range(1900, 2026, 10))
plt.xlabel("Year")
plt.ylabel("Books Read")
plt.title("Publication Years of Books I've Read")
```

Having excluded the "super-classics" that make such pronounced peaks in figure 1, this histogram shows two trends: an overall steep falloff during and before my life, and a somewhat normally distributed segment centered on the '50s. This seems to suggest that every book has a slight chance of becoming a classic after ~75 years on average.

```python
plt.hist(years, bins=range(1990, 2026, 2))
plt.xlabel("Year")
plt.ylabel("Books Read")
plt.title("Publication Years of Books I've Read")
plt.axvline(2011.5, ls="--", zorder=99, color='black')
plt.annotate("Literacy", (2011.6, 32))
plt.axvline(2004.9, ls="--", zorder=99, color='black')
plt.annotate("Birth", (2005, 32))
```

Leading up to and during my early childhood, growth appears linear, with a sharp peak around the time I learned to read, and another probably associated with late middle-school. Sharp decay for the last ten years, suggesting that most books take time to become popular.
